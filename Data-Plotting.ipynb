```python
import matplotlib.pyplot as plt

name = ["agastya","suchin","gus"]
pies = [10,20,30]

plt.pie(pies, explode = (0,0.1,0), labels = name, autopct = '%1.1f%%')
plt.title("Pie eating")
plt.xlabel("names")
plt.ylabel("# of pies")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

name = ["agastya","suchin","gus"]
pies = [10,20,30]

fig, subplots = plt.subplots(2,2)

subplots[0,0].pie(pies, explode = (0,0.1,0), labels = name, autopct = '%1.1f%%')
subplots[0,0].set_xlabel("names")
subplots[0,0].set_ylabel("# of pies")
subplots[0,0].legend(loc = "best")

subplots[0,1].bar(name,pies)
subplots[0,1].set_xlabel("names")
subplots[0,1].set_ylabel("# of pies")

subplots[1,0].plot(name,pies)
subplots[1,0].set_xlabel("names")
subplots[1,0].set_ylabel("# of pies")

names = 'Suchin'
pie = 30
subplots[1,1].scatter(names,pie)
subplots[1,1].set_xlabel("names")
subplots[1,1].set_ylabel("# of pies")

fig.tight_layout()

plt.suptitle("Subplots")
plt.show()
```

```python
import matplotlib.pyplot as plt

highfives = [12,34,56,78,90]
fistbumps = [21,43,65,87,91]
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]

width = 0.5
index = [0,1,2,3,4]

plt.style.use("ggplot")
fig = plt.subplots(figsize = (10, 4))
p1 = plt.bar(index, highfives, width)
p2 = plt.bar(index, fistbumps, width, bottom = highfives)

plt.ylabel("Number of greetings daily")
plt.xlabel("Days")
plt.title("Greetings")
plt.xticks(index,days)
plt.legend((p1[0], p2[0]), ("Highfives","Fistbumps"))

plt.show()
```

```python
import matplotlib.pyplot as plt

highfives = [12,34,56,78,90]
fistbumps = [21,43,65,87,91]
handshakes = [22,35,66,79,92]
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]

width = 0.25
index1 = [0,1,2,3,4]
index2 = []
index3 = []

for n in index1:
    index2.append(n+width)

for n in index2:
    index3.append(n+width)

plt.style.use("ggplot")
fig = plt.subplots(figsize = (10, 4))

p1 = plt.bar(index1, highfives, width)
p2 = plt.bar(index2, fistbumps, width)
p3 = plt.bar(index3, handshakes, width)

plt.ylabel("Number of greetings daily")
plt.xlabel("Days")
plt.title("Greetings")
plt.xticks(index,days)
plt.legend((p1[0], p2[0],p3[0]), ("Highfives","Fistbumps","Handshakes"))

plt.show()
```

```python
import matplotlib.pyplot as plt

x = 1
y = 2
z = 30

n = plt.axes(projection = "3d")

n.scatter3D(x, y, z, color = "blue", marker = "X")

n.set_xlabel(x, fontweight = "bold")
n.set_ylabel(y, fontweight = "bold")
n.set_zlabel(z, fontweight = "bold")

plt.show()
```

```python
import matplotlib.pyplot as plt

fig = plt.figure()
ax = fig.add_subplot(111, projection = "3d")

x = [1,2,3,4,5]
y = [8,7,6,5,4]
z = [0,0,0,0,0]

dx = [1,1,1,1,1]
dy = [1,1,1,1,1]
dz = [1,2,3,4,5]

ax.bar3d(x,y,z,dx,dy,dz, color = 'red')

ax.set_xlabel('x', fontweight = 'bold')
ax.set_ylabel('y', fontweight = 'bold')
ax.set_zlabel('z', fontweight = 'bold')
plt.title("3D bar plot")
plt.show()
```

```python
import matplotlib.pyplot as plt
from datetime import datetime

date = datetime(2019, 8, 21)
y = 10

plt.plot(date, y, 'x')

plt.gcf().autofmt_xdate()

plt.xlabel("Date")
plt.ylabel("Number of pies eaten")
plt.title("Agastya's history of pie eating", fontsize = 12)
plt.tight_layout()
plt.show()
```
